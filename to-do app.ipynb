## 2. Enter your API key securelyRun the cell and paste your key at the private prompt. `getpass` does not show the key as you type. The key stays in this runtime session only.

```python
from getpass import getpassfrom openai import OpenAI# Ask privately; do not print or save the key.api_key = getpass("Enter your OpenAI API key: ")client = OpenAI(api_key=api_key)
```

## 3. Describe an everyday objectType an object such as a paperclip, mug, or umbrella.

```python
object_name = input("Enter an everyday object: ").strip()if not object_name:    raise ValueError("Please enter an object before continuing.")
```

## 4. Ask for one creative, useful ideaThe Responses API returns one unusual idea that is practical to try.

```python
response = client.responses.create(    model="gpt-6-luna",    input=(        f"Suggest ONE unusual, creative, and practical use for a {object_name}. "        "Keep the idea safe, realistic, and concise."    ),)print(response.output_text)
```
